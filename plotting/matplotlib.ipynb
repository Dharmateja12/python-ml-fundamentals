After pd.cut(..., labels=[1,2,3,4,5]), income_category becomes a Categorical dtype — even though the labels are the integers 1–5, they're category labels, not real numeric values.

Series.hist() needs numeric data to bin, so pandas/matplotlib falls back to using the underlying category codes (0, 1, 2, 3, 4 — the positional index of each category), not your actual labels.
